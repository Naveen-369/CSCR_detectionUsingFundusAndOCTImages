```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models, transforms, datasets
from torch.utils.data import DataLoader, Dataset
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc
from sklearn.preprocessing import StandardScaler
import pennylane as qml
from pennylane import numpy as pnp

print("STEP 1: Importing libraries completed")
print("PyTorch version:", torch.__version__)

# Use GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("STEP 2: Device setup completed")
print("Using device:", device)

# Efficient dataset loading - only load target folders
print("STEP 3: Setting up efficient data loading")
data_path = "/root/.cache/kagglehub/datasets/linabennaa/original-eye-disease-image-dataset-mendeley/versions/1/Original Dataset"
target_classes = ['CSCR_Color Fundus', 'Healthy']

# Data transforms
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

print("STEP 4: Creating efficient custom dataset")
class EfficientBinaryDataset(Dataset):
    def __init__(self, root_dir, classes, transform=None):
        self.root_dir = root_dir
        self.classes = classes
        self.transform = transform
        self.samples = []
        self.class_to_idx = {classes[0]: 0, classes[1]: 1}  # CSCR=0, Healthy=1

        print("Loading samples from target folders only...")
        for class_idx, class_name in enumerate(classes):
            class_path = os.path.join(root_dir, class_name)
            if os.path.exists(class_path):
                image_files = [f for f in os.listdir(class_path)
                             if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
                print(f"Found {len(image_files)} images in {class_name}")

                for img_file in image_files:
                    img_path = os.path.join(class_path, img_file)
                    self.samples.append((img_path, class_idx))
            else:
                print(f"Warning: Path not found {class_path}")

        print(f"Total samples loaded: {len(self.samples)}")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, label = self.samples[idx]
        from PIL import Image
        image = Image.open(img_path).convert('RGB')

        if self.transform:
            image = self.transform(image)

        return image, label

# Create efficient dataset
print("STEP 5: Loading efficient binary dataset")
full_dataset = EfficientBinaryDataset(data_path, target_classes, transform)

# Split dataset
train_size = int(0.8 * len(full_dataset))
val_size = len(full_dataset) - train_size
train_dataset, val_dataset = torch.utils.data.random_split(full_dataset, [train_size, val_size])

print("STEP 6: Creating data loaders")
batch_size = 32
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)

print(f"Training samples: {len(train_dataset)}")
print(f"Validation samples: {len(val_dataset)}")
print(f"Training batches: {len(train_loader)}")
print(f"Validation batches: {len(val_loader)}")

# Quantum Setup
print("STEP 7: Setting up quantum components")
n_qubits = 4  # Number of qubits for quantum layer
dev = qml.device("default.qubit", wires=n_qubits)

def quantum_circuit(inputs, weights):
    """Quantum circuit for processing features"""
    # Data encoding - ensure float32
    for i in range(n_qubits):
        qml.RY(inputs[i].float(), wires=i)

    # Parameterized layers
    for layer in range(2):  # 2 layers
        for i in range(n_qubits):
            qml.RY(weights[layer, i, 0].float(), wires=i)
            qml.RZ(weights[layer, i, 1].float(), wires=i)

        # Entanglement
        for i in range(n_qubits - 1):
            qml.CNOT(wires=[i, i + 1])
        qml.CNOT(wires=[n_qubits - 1, 0])  # Circular entanglement

    # Measurements
    return [qml.expval(qml.PauliZ(i)) for i in range(n_qubits)]

@qml.qnode(dev, interface="torch", diff_method="parameter-shift")
def quantum_layer(inputs, weights):
    return quantum_circuit(inputs, weights)

print("STEP 8: Creating Quantum-Enhanced EfficientNet model")
class QuantumEfficientNet(nn.Module):
    def __init__(self, num_classes=2, n_qubits=4):
        super(QuantumEfficientNet, self).__init__()

        # Load pretrained EfficientNet
        self.efficientnet = models.efficientnet_b0(pretrained=True)

        # Freeze early layers for transfer learning
        for param in list(self.efficientnet.parameters())[:-10]:
            param.requires_grad = False

        # Remove the final classifier
        self.features = nn.Sequential(*list(self.efficientnet.children())[:-1])

        # Get feature dimension
        feature_dim = self.efficientnet.classifier[1].in_features

        # Classical feature reducer
        self.feature_reducer = nn.Sequential(
            nn.AdaptiveAvgPool2d((1, 1)),
            nn.Flatten(),
            nn.Linear(feature_dim, 64),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(64, n_qubits)  # Reduce to quantum input size
        )

        # Quantum layer parameters - ensure float32
        self.n_qubits = n_qubits
        self.quantum_weights = nn.Parameter(torch.randn(2, n_qubits, 2, dtype=torch.float32) * 0.1)

        # Classical post-processing
        self.classifier = nn.Sequential(
            nn.Linear(n_qubits + n_qubits, 32),  # quantum + classical features
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(32, num_classes)
        )

        print("Quantum-Enhanced EfficientNet created")

    def forward(self, x):
        # Extract classical features
        classical_features = self.features(x)
        classical_features = self.feature_reducer(classical_features)

        # Ensure float32 dtype
        classical_features = classical_features.float()

        # Process through quantum layer
        quantum_outputs = []
        for i in range(x.shape[0]):  # Process each sample
            quantum_input = classical_features[i].float()  # Ensure float32
            quantum_output = quantum_layer(quantum_input, self.quantum_weights.float())
            # Convert quantum output to tensor and ensure float32
            quantum_tensor = torch.stack([torch.tensor(q, dtype=torch.float32, device=x.device)
                                        for q in quantum_output])
            quantum_outputs.append(quantum_tensor)

        quantum_features = torch.stack(quantum_outputs).float()

        # Combine classical and quantum features
        combined_features = torch.cat([classical_features, quantum_features], dim=1)

        # Final classification
        output = self.classifier(combined_features.float())
        return output

# Initialize model
print("STEP 9: Initializing Quantum-Enhanced model")
model = QuantumEfficientNet(num_classes=2, n_qubits=n_qubits).to(device)
print("Model created and moved to device")

# Loss and optimizer
print("STEP 10: Setting up training components")
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=5, gamma=0.5)

# Training tracking
train_losses = []
train_accuracies = []
val_losses = []
val_accuracies = []

print("STEP 11: Defining training and evaluation functions")
def train_epoch(model, loader, criterion, optimizer):
    model.train()
    running_loss = 0
    correct = 0
    total = 0

    for batch_idx, (images, labels) in enumerate(loader):
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

        if batch_idx % 10 == 0:
            print(f"Batch {batch_idx}/{len(loader)} - Loss: {loss.item():.4f}")

    return running_loss / len(loader), 100 * correct / total

def evaluate_model(model, loader, criterion):
    model.eval()
    running_loss = 0
    correct = 0
    total = 0
    all_predictions = []
    all_labels = []
    all_probabilities = []

    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)

            running_loss += loss.item()
            probabilities = torch.softmax(outputs, dim=1)
            _, predicted = outputs.max(1)

            total += labels.size(0)
            correct += (predicted == labels).sum().item()

            all_predictions.extend(predicted.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            all_probabilities.extend(probabilities.cpu().numpy())

    return (running_loss / len(loader),
            100 * correct / total,
            all_predictions,
            all_labels,
            all_probabilities)

# Training loop
print("STEP 12: Starting training")
num_epochs = 10

for epoch in range(num_epochs):
    print(f"\nEPOCH {epoch + 1}/{num_epochs}")
    print("=" * 50)

    # Training
    train_loss, train_acc = train_epoch(model, train_loader, criterion, optimizer)

    # Validation
    val_loss, val_acc, _, _, _ = evaluate_model(model, val_loader, criterion)

    # Update scheduler
    scheduler.step()

    # Store metrics
    train_losses.append(train_loss)
    train_accuracies.append(train_acc)
    val_losses.append(val_loss)
    val_accuracies.append(val_acc)

    print(f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
    print(f"Val Loss: {val_loss:.4f} | Val Acc: {val_acc:.2f}%")

print("STEP 13: Training completed - Generating comprehensive evaluation")

# Final evaluation
val_loss, val_acc, predictions, true_labels, probabilities = evaluate_model(model, val_loader, criterion)
probabilities = np.array(probabilities)

print("STEP 14: Creating comprehensive plots and metrics")

# Set up plotting
plt.style.use('default')
fig = plt.figure(figsize=(20, 15))

# 1. Training History
ax1 = plt.subplot(3, 3, 1)
plt.plot(train_losses, label='Training Loss', marker='o')
plt.plot(val_losses, label='Validation Loss', marker='s')
plt.title('Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

ax2 = plt.subplot(3, 3, 2)
plt.plot(train_accuracies, label='Training Accuracy', marker='o')
plt.plot(val_accuracies, label='Validation Accuracy', marker='s')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy (%)')
plt.legend()
plt.grid(True)

# 2. Confusion Matrix
ax3 = plt.subplot(3, 3, 3)
cm = confusion_matrix(true_labels, predictions)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['CSCR', 'Healthy'],
            yticklabels=['CSCR', 'Healthy'])
plt.title('Confusion Matrix')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')

# 3. ROC Curve
ax4 = plt.subplot(3, 3, 4)
fpr, tpr, _ = roc_curve(true_labels, probabilities[:, 1])
roc_auc = auc(fpr, tpr)
plt.plot(fpr, tpr, color='darkorange', lw=2,
         label=f'ROC Curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.grid(True)

# 4. Class Distribution
ax5 = plt.subplot(3, 3, 5)
unique, counts = np.unique(true_labels, return_counts=True)
class_names = ['CSCR', 'Healthy']
plt.bar(class_names, counts, color=['skyblue', 'lightcoral'])
plt.title('Class Distribution in Validation Set')
plt.ylabel('Number of Samples')
for i, count in enumerate(counts):
    plt.text(i, count + 1, str(count), ha='center')

# 5. Prediction Confidence Distribution
ax6 = plt.subplot(3, 3, 6)
max_probs = np.max(probabilities, axis=1)
plt.hist(max_probs, bins=20, alpha=0.7, color='green')
plt.title('Prediction Confidence Distribution')
plt.xlabel('Max Probability')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# 6. Learning Rate Schedule
ax7 = plt.subplot(3, 3, 7)
lrs = [0.001 * (0.5 ** (i // 5)) for i in range(num_epochs)]
plt.plot(lrs, marker='o')
plt.title('Learning Rate Schedule')
plt.xlabel('Epoch')
plt.ylabel('Learning Rate')
plt.grid(True)

# 7. Feature Visualization (Quantum weights)
ax8 = plt.subplot(3, 3, 8)
quantum_weights = model.quantum_weights.detach().cpu().numpy()
im = plt.imshow(quantum_weights.reshape(-1, quantum_weights.shape[-1]),
                cmap='viridis', aspect='auto')
plt.title('Quantum Circuit Parameters')
plt.xlabel('Parameter Index')
plt.ylabel('Layer × Qubit')
plt.colorbar(im)

# 8. Model Performance Summary
ax9 = plt.subplot(3, 3, 9)
ax9.axis('off')
performance_text = f"""
QUANTUM-ENHANCED EFFICIENTNET RESULTS

Final Validation Accuracy: {val_acc:.2f}%
ROC AUC Score: {roc_auc:.3f}

Class-wise Performance:
CSCR Samples: {counts[0]}
Healthy Samples: {counts[1]}

Model Architecture:
- EfficientNet-B0 backbone
- {n_qubits}-qubit quantum layer
- 2-layer quantum circuit
- Combined classical-quantum features

Training Details:
- Epochs: {num_epochs}
- Batch Size: {batch_size}
- Optimizer: Adam with LR scheduling
"""
ax9.text(0.1, 0.9, performance_text, transform=ax9.transAxes,
         fontsize=10, verticalalignment='top', fontfamily='monospace')

plt.tight_layout()
plt.savefig('quantum_efficientnet_results.png', dpi=300, bbox_inches='tight')
plt.show()

# Print detailed classification report
print("\nSTEP 15: Detailed Classification Report")
print("=" * 60)
print(classification_report(true_labels, predictions,
                          target_names=['CSCR', 'Healthy'], digits=4))

# Print confusion matrix details
print("\nConfusion Matrix Analysis:")
tn, fp, fn, tp = cm.ravel()
print(f"True Negatives (CSCR correctly classified): {tn}")
print(f"False Positives (Healthy misclassified as CSCR): {fp}")
print(f"False Negatives (CSCR misclassified as Healthy): {fn}")
print(f"True Positives (Healthy correctly classified): {tp}")

# Additional metrics
sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0
specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
f1 = 2 * (precision * sensitivity) / (precision + sensitivity) if (precision + sensitivity) > 0 else 0

print(f"\nAdditional Metrics:")
print(f"Sensitivity (Recall): {sensitivity:.4f}")
print(f"Specificity: {specificity:.4f}")
print(f"Precision: {precision:.4f}")
print(f"F1-Score: {f1:.4f}")
print(f"ROC AUC: {roc_auc:.4f}")

# Save model
print("\nSTEP 16: Saving model")
torch.save(model.state_dict(), 'quantum_efficientnet_cscr_healthy.pth')
print("Quantum-Enhanced EfficientNet model saved successfully!")

print("\nTraining and evaluation completed successfully!")
print("All plots saved as 'quantum_efficientnet_results.png'")
```
